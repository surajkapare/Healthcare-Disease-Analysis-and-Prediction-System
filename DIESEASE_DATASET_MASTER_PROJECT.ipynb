## Treating outliers using the IQR method (capping) :- 


```python
Q3 = df["BMI"].quantile(0.75)
Q1 = df["BMI"].quantile(0.25)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR 

df["BMI"] = np.where(df["BMI"] > upper_bound, upper_bound, df["BMI"])
```

```python
Q3 = df["Blood_Pressure"].quantile(0.75)
Q1 = df["Blood_Pressure"].quantile(0.25)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR 

df["Blood_Pressure"] = np.where(df["Blood_Pressure"] > upper_bound, upper_bound, df["Blood_Pressure"])
```

```python
Q3 = df["Glucose_Level"].quantile(0.75)
Q1 = df["Glucose_Level"].quantile(0.25)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR 

df["Glucose_Level"] = np.where(df["Glucose_Level"] > upper_bound, upper_bound, df["Glucose_Level"])
```

```python
Q3 = df["Cholesterol"].quantile(0.75)
Q1 = df["Cholesterol"].quantile(0.25)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR 

df["Cholesterol"] = np.where(df["Cholesterol"] > upper_bound, upper_bound, df["Cholesterol"])
```

```python
Q3 = df["Heart_Rate"].quantile(0.75)
Q1 = df["Heart_Rate"].quantile(0.25)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR 

df["Heart_Rate"] = np.where(df["Heart_Rate"] > upper_bound, upper_bound, df["Heart_Rate"])
```

```python
Q3 = df["Sleep_Hours"].quantile(0.75)
Q1 = df["Sleep_Hours"].quantile(0.25)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR 
lower_bound = Q1 - 1.5 * IQR

df["Sleep_Hours"] = np.where(df["Sleep_Hours"]<lower_bound,lower_bound,df["Sleep_Hours"])
df["Sleep_Hours"] = np.where(df["Sleep_Hours"]>upper_bound, upper_bound, df["Sleep_Hours"])
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(df[num_clm])
plt.title("checking outliers after treating it")
plt.show()
```


## Problem solving on this dataset:-"

## **1. What percentage of patients are predicted to have a disease?**

```python
total = len(df)

disease_count = df["disease_prediction"].value_counts()

disease_percentage = (disease_count / total) * 100

print("Total Patients:", total)
print("\nDisease Count:")
print(disease_count)

print("\nDisease Percentage:")
print(disease_percentage.round(2))
```

## **2. What is the average health profile of disease vs non-disease patients?**

```python
cols = [
    "age",
    "bmi",
    "blood_pressure",
    "glucose_level",
    "cholesterol",
    "heart_rate",
    "sleep_hours"
]

profile = df.groupby("disease_prediction")[cols].mean().round(2)

profile
```

## **3.Which patients have multiple high-risk health indicators?**

```python
df["risk_score"] = (
    (df["bmi"] > df["bmi"].median()).astype(int) +
    (df["blood_pressure"] > df["blood_pressure"].median()).astype(int) +
    (df["glucose_level"] > df["glucose_level"].median()).astype(int) +
    (df["cholesterol"] > df["cholesterol"].median()).astype(int)
)

high_risk = df[df["risk_score"] >= 3]

print("High-risk patients:", len(high_risk))

print(
    high_risk[
        [
            "patient_id",
            "bmi",
            "blood_pressure",
            "glucose_level",
            "cholesterol",
            "risk_score"
        ]
    ].head(10)
)
```

## **4. What percentage of smokers are predicted to have disease?**

```python
smoking_disease = pd.crosstab(
    df["smoking"],
    df["disease_prediction"],
    normalize="index"
) * 100
smoking_disease.round(2)
```

## **5 Which patients have unusually high glucose levels?**

```python
Q1 = df["glucose_level"].quantile(0.25)
Q3 = df["glucose_level"].quantile(0.75)

IQR = Q3 - Q1

upper_limit = Q3 + 1.5 * IQR

high_glucose = df[df["glucose_level"] > upper_limit]

print("Upper glucose limit:", upper_limit)
print("Number of high-glucose patients:", len(high_glucose))

print(
    high_glucose[
        ["patient_id", "age", "glucose_level", "bmi", "disease_prediction"]
    ].head(10)
)
```
