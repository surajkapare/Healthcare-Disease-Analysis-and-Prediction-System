## **Project title**:- `healtcare disease analysis and prediction system.`

## **Machine Learning Model** :- `Random forest.`

## Importing the all required libraries:-

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns 

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score , confusion_matrix,recall_score,f1_score
```

## **load the dataset**:-

```python
df = pd.read_csv("health_dataset_10000_rows.csv")
df.head()
```

## **Insights**:-
- here i cleaned the data in EDA so we don't need to clean the data.

## **Encoding**:-
- we convert the categorical columns into the numeric using the label encoding.

```python
# call labelencoder:-
le = LabelEncoder()
df["sex"] = le.fit_transform(df["sex"])
df["smoking"] = le.fit_transform(df["smoking"])
df["family_history"] = le.fit_transform(df["family_history"])
df["physical_activity"] = le.fit_transform(df["physical_activity"])
```

```python
df.head()
```

## **Insights**:-
- here we converted the categorical columns into numerical using the labelencode.

## we don't need of the patient_id so we drop the patient_columnn.

```python
df = df.drop("patient_id",axis = 1)
```

```python
df.head()
```

## **defining the features and target**:-

```python
x = df.drop("disease_prediction",axis = 1)
y = df["disease_prediction"]
```

## **Insights**:-
- we defined the all features as x and the "disease_prediction" as the target 

## Spliting the features and target into train test split to train the model

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2, random_state = 46)
```

```python
## checking the splited data shape:-
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

## **Insights**:-
- so here we can see our data is splited using the `train_test_split` into training and testing section.

## **Features scaling**:-
- we scale the all features data into in one range to get best performance.

```python
## calling the scaler:- 
scaler = StandardScaler()
x_train_scl = scaler.fit_transform(x_train)
x_test_scl = scaler.fit_transform(x_test)
```

```python
x_train_scl
```

```python
x_test_scl
```

## **Insights**:- 
- here we can see the of my data is scaled into the one range.

## **Training model**:- (`Random Forest`)

```python
random_f = RandomForestClassifier(
    n_estimators=200,
    class_weight='balanced',
    max_depth=10,
    min_samples_leaf=5,
    random_state=42)
random_f.fit(x_train_scl,y_train)
```

## Making prediction:-

```python
pred = random_f.predict(x_test_scl)
comp = pd.DataFrame({
    "actual prediction" : y_test,
    "model prediction" : pred
})
comp
```

## **Insights**:-
- we trained the model.
- we make prediction and compared it with actual values.

## **Evaluation**:-

```python
## calling all parameters to evaluate the model(accuracy_score,confusion_matrix):-
Accuracy = accuracy_score(y_test,pred)
cm = confusion_matrix(y_test,pred)
```

```python
print(Accuracy*100,"%")
print(cm)
```

## **Insights**:- 
- here we can see the our model is giving the `94.5%` accuracy which is best.
